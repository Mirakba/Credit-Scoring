```python
query = """
WITH params AS (
    SELECT DATE '2025-04-30' AS as_of_date
),
payment_stats AS (
SELECT
    c.application_id,
    -- Максимальная просрочка за первые 6 месяцев
    MAX(
        GREATEST(
            CASE
                WHEN p.paid_date IS NULL THEN
                    LEAST(
                        (SELECT as_of_date FROM params),
                        c.issue_date + INTERVAL '6 months'
                    )::date - p.due_date
                ELSE
                    p.paid_date - p.due_date
            END,
            0
        )
    ) AS max_dpd_6m,
    -- Суммарное количество дней просрочки
    SUM(
        GREATEST(
            CASE
                WHEN p.paid_date IS NULL THEN
                    LEAST(
                        (SELECT as_of_date FROM params),
                        c.issue_date + INTERVAL '6 months'
                    )::date - p.due_date
                ELSE
                    p.paid_date - p.due_date
            END,
            0
        )
    ) AS late_days_6m,
    -- Количество просроченных платежей
    COUNT(*) FILTER (
        WHERE
            GREATEST(
                CASE
                    WHEN p.paid_date IS NULL THEN
                        LEAST(
                            (SELECT as_of_date FROM params),
                            c.issue_date + INTERVAL '6 months'
                        )::date - p.due_date
                    ELSE
                        p.paid_date - p.due_date
                END,
                0
            ) > 0
    ) AS late_payments_6m
FROM scoring_contracts c
JOIN scoring_payments p
    ON c.contract_id = p.contract_id
CROSS JOIN params
WHERE
    p.due_date <= c.issue_date + INTERVAL '6 months'
    -- оставляем только зрелые кредиты
    AND c.issue_date <= params.as_of_date - INTERVAL '6 months'
GROUP BY c.application_id
)
SELECT
    a.*,
    ps.max_dpd_6m,
    ps.late_days_6m,
    ps.late_payments_6m,
    CASE
        WHEN a.final_decision = 'rejected' THEN 1
        WHEN ps.max_dpd_6m >= 90 THEN 1
        ELSE 0
    END AS target
FROM scoring_applications a
LEFT JOIN payment_stats ps
    ON a.application_id = ps.application_id
CROSS JOIN params
WHERE
    a.decision_date <= params.as_of_date - INTERVAL '6 months';
"""

df7 = pd.read_sql(query, engine)
```

```python
def dispute_features(application):

    client = application.client_id
    decision = application.decision_date

    disp = df_disputes[
        (df_disputes['client_id'] == client) &
        (df_disputes['opened_date'] <= decision) &
        (df_disputes['opened_date'] >= decision - relativedelta(months=6))
    ]

    if disp.empty:
        return pd.Series({
            'dispute_cnt_6m': 0,
            'dispute_amount_6m': 0,
            'fraud_claim_cnt_6m': 0,
            
        })

    return pd.Series({

        # всего диспутов
        'dispute_cnt_6m':
            len(disp),

        # общая сумма диспутов
        'dispute_amount_6m':
            disp['dispute_amount_uzs'].sum(),

        # причины диспутов
        'fraud_claim_cnt_6m':
            (disp['reason_code'] == 'fraud_claim').sum(),
    })
```

```python
import pandas as pd

def add_contract_features(df, contracts_df):

    def contract_features(application):

        client_id = application["client_id"]
        decision_date = application["decision_date"]

        contracts = contracts_df[
            (contracts_df["client_id"] == client_id) &
            (contracts_df["issue_date"] <= decision_date)
        ].copy()

        if contracts.empty:
            return pd.Series({
                "existing_monthly_payment": 0,
                "contract_cnt": 0,
                "total_debt": 0,
                "avg_rate": 0
            })

        contracts["expected_end_date"] = [
            d + pd.DateOffset(months=int(m))
            for d, m in zip(
                contracts["issue_date"],
                contracts["term_months"]
            )
        ]

        active_contracts = contracts[
            contracts["expected_end_date"] > decision_date
        ]

        return pd.Series({
            "existing_monthly_payment": active_contracts["monthly_payment_uzs"].sum(),
            "contract_cnt": len(active_contracts),
            "total_debt": active_contracts["issued_amount_uzs"].sum(),
            "avg_rate": active_contracts["annual_rate_pct"].mean() if not active_contracts.empty else 0
        })

    features = df.apply(contract_features, axis=1)

    return pd.concat([df, features], axis=1)
```

```python

df_last = df_last.sort_values(["client_id", "decision_date"]).copy()

# 3. Создаем флаг ДО заполнения
df_last["employment_missing"] = df_last["employment_months"].isna().astype(int)


# 4. Заполняем из предыдущих заявок
def fill_from_history(group):
    last_emp = None
    last_date = None

    for idx, row in group.iterrows():

        # если стаж известен — обновляем последнее значение
        if pd.notna(row["employment_months"]):
            last_emp = row["employment_months"]
            last_date = row["decision_date"]

        # если стаж отсутствует, но есть предыдущая заявка
        elif last_emp is not None:

            months = (
                (row["decision_date"].year - last_date.year) * 12 +
                (row["decision_date"].month - last_date.month)
            )

            group.at[idx, "employment_months"] = last_emp + months

    return group


df_last = (
    df_last.groupby("client_id", group_keys=False)
      .apply(fill_from_history)
)

# 5. Заполняем оставшиеся NaN медианой
median_emp = df_last["employment_months"].median()

df_last["employment_months"] = (
    df_last["employment_months"]
      .fillna(median_emp)
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

# Предобработка
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Pipeline
pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=2000,
        random_state=42
    ))
])

# Гиперпараметры
param_grid_lr = {
    "classifier__C": [0.0001,0.01, 0.1, 1, 10],
    "classifier__penalty": ["l1", "l2",'elasticnet'],
    "classifier__solver": ["liblinear",'saga',]
}

# GridSearchCV
grid_lr = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid_lr,
    scoring="roc_auc",
    cv=tscv,
    n_jobs=-1,
    verbose=2
)

# Обучение
grid_lr.fit(X_train, y_train)

print("Best params:", grid_lr.best_params_)
print("Best ROC-AUC:", grid_lr.best_score_)
```

```python
from sklearn.metrics import (
    roc_auc_score,
    log_loss,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    average_precision_score,
    brier_score_loss,
    confusion_matrix
)

def evaluate(name, y_true, pred, proba):

    print(f"\n{name}")

    print("ROC-AUC:", roc_auc_score(y_true, proba))
    print("PR-AUC:", average_precision_score(y_true, proba))
    print("LogLoss:", log_loss(y_true, proba))
    print("Brier:", brier_score_loss(y_true, proba))

    print("Accuracy:", accuracy_score(y_true, pred))
    print("Balanced Accuracy:", balanced_accuracy_score(y_true, pred))
    print("Precision:", precision_score(y_true, pred))
    print("Recall:", recall_score(y_true, pred))
    print("F1:", f1_score(y_true, pred))
    print("MCC:", matthews_corrcoef(y_true, pred))

    print("\nConfusion matrix")
    print(confusion_matrix(y_true, pred))
    
    print("Best iteration:", lgb_final.best_iteration_)
    print("Train AUC:", roc_auc_score(y_train, proba_train))
    print("Validation AUC:", roc_auc_score(y_valid, proba_valid))
    print("Gap:", roc_auc_score(y_train, proba_train) - roc_auc_score(y_valid, proba_valid))
```

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    confusion_matrix
)

thresholds = [i/100 for i in range(10, 95, 5)]  # от 0.10 до 0.90

results = []

for t in thresholds:

    pred = (proba_valid >= t).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_valid, pred).ravel()

    specificity = tn / (tn + fp)
    approval_rate = (pred == 0).mean()   # 0 = одобряем, 1 = дефолт

    results.append({
        "Threshold": t,
        "Accuracy": accuracy_score(y_valid, pred),
        "Precision": precision_score(y_valid, pred, zero_division=0),
        "Recall": recall_score(y_valid, pred),
        "F1": f1_score(y_valid, pred),
        "Specificity": specificity,
        "Balanced Acc": balanced_accuracy_score(y_valid, pred),
        "MCC": matthews_corrcoef(y_valid, pred),
        "Approval Rate": approval_rate
    })

results = pd.DataFrame(results)

print(results.round(4))
```
